Manual Challenges for IMC Prosperity 3

R1: You're given different currencies and need to find a sequence of trades to obtain the maximum possible amount from arbitrage.

Solution: Create a 2D array with all of the rates, then run a single nested loop structure to determine the best answer. You don't have to worry about the possibility of performing fewer trades; these are accounted for by picking a 1 on the diagonal. If you need to exclude these, add another if statement to check inside the nested loop structure.

R2: You're given a selection of crates with inhabitants, and you and the other participants are going to select up to K crates, for some postive integer K. For each crate, the value you obtain from it is the quotient of its total value and the sum (# inhabitants + % picked). For instance, (5 inhab + 10% picked) = 15 pieces divided into. You also have fees for the more crates you pick.

Solution: First determine the Nash distribution (findable by binary search or gradient descent) and how many boxes are optimal (so long as the expected gain from a box as computed by the Nash exceeds the fee, take it). Then determine how chat will react by using level 1 thinking (obvious first step) and play elsewhere. The center numbers are usually very good, since the Nash at the high and low ends are fairly low %, which means high risk.

R3: You're trading with turtles who have prices in some distribution. Each turtle will trade its item with you if you have the lowest price above its already-set reserve price. However, don't go too low, as then there is less chance that the turtles will trade with you. Since the items are of high demand, you'll be able to sell each of the items for the max possible after the round is over; the question is how much you are willing to undercut the turtles.

Solution: Hard-code the cumulative density funciton and the expected profit from just your two guesses. Note that your lower guess will take away some probability of trading from your higher one. Then, iterate through all options to find the best one. You may have to change one or both of your guesses slightly to accommodate any catches found in the question.

R5: You're given a news article, and you're asked to determine how different products will react in the market. You can put a percentage of your total capital into each item, but beware that the more you invest, the higher the trading fee for that item will be.

Solution: Put an extremely high sell on products guaranteed to go down. Don't do anything with the rest; they're unpredictable due to variance. Keep in mind that you're buying the actual products *themselves*, not the stocks! There's also a rather large fee for putting everything into one asset, so split among 3 to 4 if possible. You should also calculate how much you should put in each separately to maximize that product's profit - this can be done by hand or by graphing the PNL as a function of % invested.
